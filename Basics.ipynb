Q.1 Write a program that takes your name, age, and city as input on a single line (space-separated) and
prints a formatted sentence introducing yourself using an f-string.

```python
def intro(info):
  words=info.split()
  name=words[0]
  age=words[1]
  city=" ".join(words[2:])
  print(f"Hello , I am {name} . I am {age} years old and I live in {city}!")


info = input("enter Your name, age and city : ")
intro(info)
```

Q.2 Take the radius of a circle as input and print its area and circumference, both rounded to 2 decimal
places.

```python

def area(radius):
  area=3.1415*(radius**2)
  circumference=2*3.1415*radius
  print(f"Area : {area:.2f}, Circumference : {circumference:.2f}")

radius=int(input("Enter Radius : "))
area(radius)
```

Q4. Write a program to swap two numbers WITHOUT using a third variable, using two different methods
(arithmetic and Pythonic tuple swap).


```python
##Pythonic Tuple
num1= 10
num2 = 20

num1,num2=num2,num1

print(f"Number 1 : {num1} \nNumber 2 : {num2}")
```

```python
##Addition and Subtraction
num1= 10
num2= 20
sum=num1+num2
num1=sum-num1
num2=sum-num2
print(f"Number 1 : {num1} \nNumber 2 : {num2}")
```

```python
##Multiplication and Division
a = 5
b = 10

# Swap
a = a * b    # a becomes 50
b = a // b   # b becomes 50 // 10 = 5 (Using // for integer division)
a = a // b   # a becomes 50 // 5 = 10

print(f"Multiplication/Division: a = {a}, b = {b}")
```

```python
##Bitwise XOR (Integers Only)
a = 5  # Binary: 0101
b = 10 # Binary: 1010

# Swap
a = a ^ b  # a becomes 1111 (15)
b = a ^ b  # b becomes 0101 (5)
a = a ^ b  # a becomes 1010 (10)

print(f"Bitwise XOR: a = {a}, b = {b}")
```
