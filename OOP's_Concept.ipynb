1. 🎓 Student Result Comparator
Problem Definition
Create a Student class containing:

Student name

Roll number

Marks

Create two student objects.

The program should allow the two student objects to be compared based on their marks.

The program should determine:
Which student scored higher
Which student scored lower
Whether both students scored the same marks
The comparison should be performed directly between the objects.


```python
class Student:

  def __init__(self,name,roll_number,marks):
    self.name = name
    self.roll_number = roll_number
    self.marks = marks

  def __gt__(self,other):
    return self.marks > other.marks

  def __eq__(self,other):
    return self.marks == other.marks

  def __lt__(self,other):
    return self.marks < other.marks

  def display(self,label):
    print(f"{label}: {self.name}")
    print(f"Marks : {self.marks}")



s1 = Student("Amit", 101, 85)
s2 = Student("Rahul", 102, 78)

s1.display("Student 1")
s2.display("Student 2")

if s1 > s2:
  print(f"Result: {s1.name} has Greater Marks ")
elif s1 < s2:
  print(f"Result: {s2.name} has Greater Marks ")
elif s1 == s2:
  print(f"Result: {s1.name} and {s2.name} has Equal Marks ")


```

2. 🚗 Vehicle Showroom
Problem Definition
Create a program for a vehicle showroom.
The showroom sells:

Cars

Bikes

Trucks

All vehicles should have common information such as brand and model.
However, each type of vehicle should provide its own way of displaying its details.

The main program should be able to work with different types of vehicles without needing separate logic for every vehicle type.


```python
class Vehicle:

  def __init__(self, brand: str, model: str):
    self.brand = brand
    self.model = model

  def display_details(self):
    pass


class Car(Vehicle):

  def display_details(self):
    print("Car")
    print(f"Brand: {self.brand}")
    print(f"Model: {self.model}\n")


class Bike(Vehicle):

  def display_details(self):
    print("Bike")
    print(f"Brand: {self.brand}")
    print(f"Model: {self.model}\n")


class Truck(Vehicle):

  def display_details(self):
    print("Truck")
    print(f"Brand: {self.brand}")
    print(f"Model: {self.model}\n")


  showroom = [
      Car("Toyota", "Innova"),
      Bike("Honda", "Activa"),
      Truck("Tata", "407"),
  ]

  print("----- Vehicle Details -----\n")
  for vehicle in showroom:
    vehicle.display_details()
```

3. 🍕 Pizza Bill Calculator
Problem Definition
Create a pizza ordering system.

The system should support different types of pizzas:

Regular Pizza

Cheese Pizza

Premium Pizza

Each pizza should have a different way of calculating its final price.
The program should accept pizza objects and calculate their final prices.
The system should be designed so that another pizza type can be added in the future without changing the existing calculation structure significantly.


```python
class Pizza:
  def __init__(self,base_price):
    self.base_price=base_price

  def calculate_price(self):
    return self.base_price

  def display_details(self):
    print(f"Regular Pizza Price : {self.calculate_price()}")


class Regular(Pizza):
  pass

class Cheese(Pizza):

  def __init__(self,base_price,cheese_cost):
    super().__init__(base_price)
    self.cheese_cost = cheese_cost

  def calculate_price(self):
    return self.base_price + self.cheese_cost

  def display_details(self):
    print(f"Cheese Pizza Price : {self.calculate_price()}")

class Premium(Pizza):

  def __init__(self,base_price,toppings):
    super().__init__(base_price)
    self.toppings= toppings

  def calculate_price(self):
    return self.base_price + self.toppings

  def display_details(self):
    print(f"Premium Pizza price : {self.calculate_price()}")


order = [
      Regular(base_price=200),
      Cheese(base_price=200, cheese_cost=80),
      Premium(base_price=200, toppings=150),
  ]


for pizza in order:
  pizza.display_details()




```

5. 🏦 Digital Wallet
Problem Definition
Create a DigitalWallet class.

The wallet should contain:

Owner name

Wallet ID

Balance

The balance should not be directly modified from outside the class.

Provide operations for:

Adding money

Spending money

Checking balance

The program should prevent invalid operations such as spending more money than the available balance.


```python
class DigitalWallet:
  def __init__(self,owner,wallet_id,initial_balance):
    self.owner = owner
    self.wallet_id= wallet_id
    self.__balance = initial_balance if initial_balance >= 0 else 0.0

  def get_balance(self):
    return self.__balance

  def add_money(self, amount):
    if amount<= 0:
      print("Error : Deposit amount Must be Positive. ")
      return False
    self.__balance += amount
    print(f"Money Added: {amount}")
    print(f"Current Balance: {self.__balance}\n")
    return True

  def spend_money(self, amount: float) -> bool:
    if amount <= 0:
      print("Error: Spending amount must be positive.")
      return False

    if amount > self.__balance:
      print(f"Error : Insufficiant Balance ! Available: {self.__balance}")
      return False
    self.__balance -= amount
    print(f"Money Spent : {amount}")
    print(f"Final Balance : {self.__balance}\n")
    return True


wallet = DigitalWallet(owner ="Prasad",wallet_id="W101",initial_balance=5000)

print(f"Owner : {wallet.owner}")
print(f"Initial Balance : {wallet.get_balance()}\n")

wallet.add_money(2000)
wallet.spend_money(1500)
```

7. 🧮 Smart Calculator Objects
Problem Definition
Create a class named Number.

Each object should contain one numerical value.

The program should allow two Number objects to be added directly.

It should also allow:

Addition

Subtraction

between objects.

The result should be another Number object


```python
class Number:
  def __init__(self,value):
    self.value=value

  def __add__(self,other):
    if isinstance(other,Number):
      return Number(self.value + other.value)
    return NotImplemented


  def __sub__(self,other):
    if isinstance(other,Number):
      return Number(self.value-other.value)
    return NotImplemented


  def __str__(self):
    return (
        f"{int(self.value)}"
        if isinstance(self.value,(int,float)) and self.value.is_integer()
        else f"{self.value}"
    )


num1 =Number(50)
num2 =Number(20)

print(f"Number 1 : {num1}")
print(f"Number 2 : {num2}\n")

sum_result = num1 +num2
sub_result =num1-num2

print(f"Addition : {sum_result}")
print(f"Substraction : {sub_result}")
```

8. 💳 Payment Gateway
Problem Definition
Design a payment system supporting:

UPI

Credit Card

Net Banking

Every payment method should provide a common operation for processing payment.

However, the actual payment processing behavior should be different for each payment method.

The main program should be able to process different payment types without knowing their internal implementation.


```python
class PaymentMethod:

  def process_payment(self, amount: float):
    pass


class UPI(PaymentMethod):

  def __init__(self, upi_id: str):
    self.upi_id = upi_id

  def process_payment(self, amount: float):
    print(f"Payment Amount: {amount}")
    print("Payment Method: UPI")
    print(f"UPI ID: {self.upi_id}\n")
    print("Payment processed successfully using UPI.")


class CreditCard(PaymentMethod):

  def __init__(self, card_number: str, card_holder: str):
    self.card_number = f"**** **** **** {card_number[-4:]}"
    self.card_holder = card_holder

  def process_payment(self, amount: float):
    print(f"Payment Amount: {amount}")
    print("Payment Method: Credit Card")
    print(f"Card Holder: {self.card_holder}")
    print(f"Card Number: {self.card_number}\n")
    print("Payment processed successfully using Credit Card.")


class NetBanking(PaymentMethod):

  def __init__(self, bank_name: str, account_number: str):
    self.bank_name = bank_name
    self.account_number = account_number

  def process_payment(self, amount: float):
    print(f"Payment Amount: {amount}")
    print("Payment Method: Net Banking")
    print(f"Bank Name: {self.bank_name}\n")
    print("Payment processed successfully using Net Banking.")


def process_checkout(payment_method: PaymentMethod, amount: float):
  payment_method.process_payment(amount)

upi_payment = UPI(upi_id="Prasad@upi")
process_checkout(payment_method=upi_payment, amount=5000)
```

9. 🏠 Smart Home Devices
Problem Definition
Create a smart-home system containing different devices:

Fan

Light

Air Conditioner

Every device should support operations to:

Turn ON

Turn OFF

However, each device may perform the operation differently.

Create a common structure for smart devices and implement the required behavior for each device.


```python
class SmartDevice:

  def __init__(self, device_name: str):
    self.device_name = device_name
    self.is_on = False

  def turn_on(self):
    self.is_on = True
    print(f"{self.device_name} turned ON.")

  def turn_off(self):
    self.is_on = False
    print(f"{self.device_name} turned OFF.")


class Fan(SmartDevice):

  def __init__(self):
    super().__init__("Fan")

  def turn_on(self):
    self.is_on = True
    print("Fan turned ON.")

  def turn_off(self):
    self.is_on = False
    print("Fan turned OFF.")


class Light(SmartDevice):

  def __init__(self):
    super().__init__("Light")

  def turn_on(self):
    self.is_on = True
    print("Light turned ON.")

  def turn_off(self):
    self.is_on = False
    print("Light turned OFF.")



class AirConditioner(SmartDevice):

  def __init__(self):
    super().__init__("Air Conditioner")

  def turn_on(self):
    self.is_on = True
    print("Air Conditioner turned ON.")

  def turn_off(self):
    self.is_on = False
    print("Air Conditioner turned OFF.")


fan = Fan()
light = Light()
ac = AirConditioner()

fan.turn_on()
light.turn_on()
ac.turn_off()
```
