Q1. Reverse a String

Write a function reverse_string(s) that returns the reverse of the given string without using the built-in reverse
helpers on lists as the only solution. Explain slicing as an accepted interview approach.

```python
def reverse_string(s: str)->str:
  char_list =list(s)
  left,right = 0,len(char_list)-1


  while left < right:
    char_list[left] , char_list[right] = char_list[right] , char_list[left]
    left+=1
    right-=1

  return "".join(char_list)

reverse_string("abcdefg")
```

Q2. Check Palindrome (Ignore Case and Spaces)

Write a function is_palindrome(s) that returns True if s reads the same forwards and backwards after removing
spaces and ignoring case.

```python
def is_palindrome(s):
  blank=s.replace(" ","").lower()
  return (f"Yes [{s}] is palindrome") if blank[::]== blank[::-1] else (f"No [{s}] is Not palindrome")

is_palindrome("A man a plan a canal Panama")
```

Q.3 Write a function count_letters(s) that returns a dictionary with keys 'vowels' and 'consonants' counting only
alphabetic characters. Ignore case and skip digits/symbols.

```python
def count_letters(s):
  letters = s.replace(" ","").lower()
  dict_letters={"vowels": 0, "Consonants": 0,"digit":0,"symbol":0}
  for i in letters:
    if i in "aeiou":
      dict_letters["vowels"]+=1
    elif i.isnumeric():
      dict_letters["digit"]+=1
    elif i.isalpha():
      dict_letters["Consonants"]+=1
    else:
      dict_letters["symbol"]+=1
  return dict_letters

print(count_letters("aBC DEgj enmia  @3#^&% adjKL AHXVTA9375 4565 15  UBAIOI OIYU "))
```

Q.5 Write a function first_unique_char(s) that returns the first character that appears exactly once. Return None if every
character repeats.


```python
def first_unique_char(s:str):
    count={ }
    for char in s:
        count[char]= count.get(char, 0) + 1

    for char in s:
        if count[char] == 1:
            return char

    return None

print(first_unique_char("leetcode"))
print(first_unique_char("loveleetcode"))
print(first_unique_char("aabbccza"))



```

Q.6 Write a function are_anagrams(s1, s2) that returns True if two strings contain the same characters with the same
frequencies (ignore case and spaces).

```python
def are_anagrams(s1:str,s2:str) -> bool:
    clean_s1=s1.replace(" ","").lower()
    clean_s2=s2.replace(" ","").lower()

    if len(clean_s1)!=len(clean_s2):
        return False

    def get_frequencies(s):
        counts={}
        for char in s:
            counts[char]=counts.get(char, 0)+1
        return counts

    return get_frequencies(clean_s1)==get_frequencies(clean_s2)

print(are_anagrams("Listen","Silent"))
print(are_anagrams("Astronomer","Moon starer"))
```

Q.8 Write a function second_largest(nums) that returns the second largest distinct value. Return None if the list has fewer
than two distinct numbers.


```python
def second_largest(nums:list)->int:
  largest= 0
  second= 0
  for num in nums:
    if num > largest:
      second=largest
      largest = num
    else:
      if num > second:
        second = num

  return second

a=[1, 2, 3, 4, 5, 16, 17, 8, 9,10,12,15]
second_largest(a)
```

Q.9 Write a function rotate_right(nums, k) that rotates the list to the right by k steps. k may be larger than the list length.

```python
def rotate_right(nums, k):
    if not nums:
        return []

    k %=len(nums)
    if k == 0:
        return nums[:]

    return nums[-k:]+nums[:-k]

nums = [1, 2, 3, 4, 5, 6, 7, 8]
k= 2
print(rotate_right(nums, k))
```

Q.10 Write a function pair_with_sum(nums, target) that returns the first pair of values that add up to target. Return None
if no such pair exists. Each element may be used at most once.

```python
def pair_with_sum(nums, target):
    seen = set()

    for num in nums:
        complement = target - num
        if complement in seen:
            return (complement, num)
        seen.add(num)

    return None

nums = [2,7,11,15]
target=9
print(pair_with_sum(nums, target))
```

Q.12 Write a function group_anagrams(words) that groups words that are anagrams of each other and returns a list of
groups

```python
def group_anagrams(words):
    anagrams ={}

    for word in words:
        sorted_signature= ''.join(sorted(word))

        if sorted_signature not in anagrams:
            anagrams[sorted_signature]= []
        anagrams[sorted_signature].append(word)

    return list(anagrams.values())

words = ["eat", "tea", "tan", "ate", "nat", "bat"]
print(group_anagrams(words))
```

Q.13 Write a function flatten(items) that flattens a nested list of arbitrary depth into a single list of values

```python
def flatten(items):
    result=[]

    for item in items:
        if isinstance(item, list):
            result.extend(flatten(item))
        else:
            result.append(item)
    return result

items = [1, [2, [3, 4], 5], 6]
print(flatten(items))
```

Q.14 Write a function merge_sum(d1, d2) that merges two dictionaries. If a key exists in both, store the sum of the values.
Do not mutate the inputs.

```python
def merge_sum(d1, d2):
    result =d1.copy()

    for key,value in d2.items():
        result[key] =result.get(key, 0)+ value

    return result

d1 = {'a': 1,'b': 20,'c': 5}
d2 = {'b': 10,'c':15,'d': 7}
print(merge_sum(d1,d2))
```

Q.15 You are given a dictionary mapping student names to a list of marks. Write a function toppers(scores) that returns a
list of names whose average mark is strictly greater than the class average. Sort names alphabetically.


```python
def toppers(scores):
  student_averages= {}
  total_marks_sum= 0
  total_marks_count= 0

  for student, marks in scores.items():
    if marks:
        student_avg =sum(marks) / len(marks)
        total_marks_sum +=sum(marks)
        total_marks_count+= len(marks)
    else:
        student_avg = 0
    student_averages[student]= student_avg

  if total_marks_count== 0:
    return []
  class_average =total_marks_sum/total_marks_count

  top_students =[student for student, avg in student_averages.items() if avg > class_average]

  return sorted(top_students)

scores = {
    'Asha':[80, 90, 70],
    'Rohan':[60, 65, 55],
    'Meera':[90, 95, 85],
    'Dev':[70, 70, 70]
}
print(toppers(scores))
```
