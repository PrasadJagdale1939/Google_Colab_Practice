Q.7 Write a Python program that takes a list of strings and returns a new list containing only
the palindromes.

```python
strings_list = ["racecar", "hello", "radar", "computer", "level", "python", "kayak", "world", "madam", "apple", "noon", "guitar"]
palindrome=[]
non_palindrome=[]

for i in strings_list:
  if i==i[::-1]:
    palindrome.append(i)
  else:
    non_palindrome.append(i)

print(f"Palindrome are {palindrome}")
print(f"Non Palindrome are {non_palindrome}")

```

Q.15 Write a Python program that takes a list of integers and returns a new list containing only
the prime numbers

```python
list1=[1, 2, 3, 4, 5, 10, 13, 15, 17, 20, 23]

primes=[]
for n in list1:
    if n > 1:
        is_prime = True

        for i in range(2, n):
            if n % i == 0:
                is_prime = False
                break

        if is_prime:
            primes.append(n)

print(primes)
```

Q.21 Write a Python program that takes a list of stock prices and finds the percentage
increase/decrease from first to last day.

```python
stock_prices = [150.50, 152.10, 151.80, 149.50, 161.50,145.7]

if len(stock_prices) >= 2:
    first_price = stock_prices[0]
    last_price = stock_prices[-1]

    change = ((last_price - first_price) / first_price) * 100

    print(f"First day: ${first_price:.2f}")
    print(f"Last day:  ${last_price:.2f}")

    if change > 0:
        print(f"Result: Increase of {change:.2f}%")
    elif change < 0:
        print(f"Result: Decrease of {abs(change):.2f}%")
    else:
        print("Result: No change (0.00%)")
else:
    print("Not enough data to calculate percentage change.")
```
