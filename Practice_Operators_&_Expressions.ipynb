Q.5 Predict the output of: print(10 > 5 and 3 < 1 or not False) . Explain the operator precedence
used

```python
print(10 > 5 and 3 < 1 or not False)
##True
```

Original: 10 > 5 and 3 < 1 or not False

Step 1: Evaluate the comparisons

10 > 5 is True

3 < 1 is False

The equation is now: True and False or not False

Step 2: Evaluate not

not False is True

The equation is now: True and False or True

Step 3: Evaluate and

True and False is False (both sides must be True for and to be True)

The equation is now: False or True

Step 4: Evaluate or

False or True is True (only one side needs to be True for or to be True)
