Q.1 Write a program to classify a given year as a leap year or not

```python
year=int(input("Enter Year : "))

if (year%4==0 and year%100!=0) or (year%400==0):
  print("yes It is Leap Year !")
else:
  print("No It's Not Leap Year !")
```

Q.2 Write a BMI calculator: take height (m) and weight (kg), calculate BMI, and print a category
(Underweight/Normal/Overweight/Obese) using if-elif-else.


```python
height, weight =map(float,input("Enter Your Height(m) and weight(kg) : ").split())

BMI=weight/(height**2)
if BMI < 18.5:
  print(f"Your is {BMI:.2f} you are Under Underweight")
elif BMI < 25:
  print(f"Your is {BMI:.2f} you are Under Normal")
elif BMI < 30 :
  print(f"Your is {BMI:.2f} you are Under Overweight")
else:
  print(f"Your is {BMI:.2f} you are Under Obese")
```

Q.3 Write a program that takes three sides of a triangle and determines if they form a valid triangle, and if so,
whether it's equilateral, isosceles, or scalene.


```python
a,b,c =map(float,input("Enter Your length of Sides of Triangle : ").split())

if a==b==c:
  print(f"It is a Equilateral Triangle")
elif a==b!=c or a==b or b==c or a==c:
  print(f"It is a Isosceles Triangle")
else:
  print(f"It is a scalene Triangle")

```

Q.4 Write a simple grading system using match-case that converts a percentage into a letter grade (A+/A/
B/C/Fail).


```python
marks=int(input("Enter Your Marks : "))
match marks:
    case _ if marks > 85:
        print("+A")
    case _ if marks > 70:
        print("A")
    case _ if marks > 60:
        print("B")
    case _ if marks >= 40:
        print("C")
    case _:
        print("Fail")

```

Q.5 Write a program to check whether a given character is a vowel, consonant, digit, or special symbol.

```python
def check(string: str)->str:
  cleaned=string.strip().lower().replace(" ","")

  dictionary ={}

  for letter in cleaned:
    if letter.isalpha():
      if letter in "aeiou":
        dictionary["vowels"]=dictionary.get("vowels",0)+1
      else:
        dictionary["Consonant"]=dictionary.get("Consonant",0)+1
    elif letter.isnumeric():
      dictionary["Digit"]=dictionary.get("Digit",0)+1
    else:
      dictionary["Symbol"]=dictionary.get("Symbol",0)+1

  return dictionary

result=check("avag bjb cvhA SAHJ DAH AGIU bd dgu15 162 5252  87 812 658 68237 8!## @#$  %^&  *(*&^  %$#@    $%^ &*)")
print(result)
```

Q.6 Write a program that takes a day number (1–7) and prints the day name using match-case, with a
default case for invalid input.


```python
day = int(input("Enter the day number : "))
match day:
  case 1:
    print("Monday")
  case 2:
    print("Tuesday")
  case 3:
    print("Wednesday")
  case 4:
    print("Thursday")
  case 5:
    print("Friday")
  case 6:
    print("Saturday")
  case 7:
    print("Sunday")
  case _:
    print("Invalid Input Try Between (1 - 7)")
```

Q.7 Write a program to determine the quadrant of a point (x, y) on a Cartesian plane (or state if it lies on an
axis).


```python
def point(x,y:tuple):
  if x>0 and y>0:
    print(f"{x} and {y} is in I Quadrant")
  if x<0 and y>0:
    print(f"{x} and {y} is in II Quadrant")
  if x<0 and y<0:
    print(f"{x} and {y} is in III Quadrant")
  if x>0 and y<0:
    print(f"{x} and {y} is in IV Quadrant")

a,b=1,-1
point(a,b)
```
