**Q1.** Compress a string by counting consecutive repeated characters (e.g. "aaabbc" -> "a3b2c1").

```python
def compress_string(s):
    # Handle empty string
    if not s:
        return ""
    result = []
    count = 1
    for i in range(1, len(s)):
        if s[i] == s[i - 1]:
            count += 1
        else:
            result.append(s[i - 1] + str(count))
            count = 1
    result.append(s[-1] + str(count))  # append last group
    return "".join(result)

print(compress_string("aaabbc"))
print(compress_string("aabcccccaaa"))
```

**Q2.** Check if a string can be rearranged to form a palindrome.

```python
def can_form_palindrome(s):
    # Count frequency of each character
    freq = {}
    for ch in s:
        freq[ch] = freq.get(ch, 0) + 1
    # A string can form a palindrome if at most one character has an odd count
    odd_count = sum(1 for v in freq.values() if v % 2 != 0)
    return odd_count <= 1

print(can_form_palindrome("civic"))
print(can_form_palindrome("ivicc"))
print(can_form_palindrome("hello"))
```

**Q3.** Find the first non-repeating character in a string.

```python
def first_non_repeating(s):
    freq = {}
    for ch in s:
        freq[ch] = freq.get(ch, 0) + 1
    for ch in s:
        if freq[ch] == 1:
            return ch
    return None

print(first_non_repeating("swiss"))
print(first_non_repeating("aabbcc"))
```

**Q4.** Check if one string is a rotation of another (e.g. "abcd" and "cdab").

```python
def is_rotation(s1, s2):
    # A string s2 is a rotation of s1 if s2 is a substring of s1+s1 and lengths match
    if len(s1) != len(s2):
        return False
    return s2 in (s1 + s1)

print(is_rotation("abcd", "cdab"))
print(is_rotation("abcd", "acbd"))
```

**Q10.** Print a spiral pattern of numbers in an n x n grid.

```python
def spiral_pattern(n):
    matrix = [[0] * n for _ in range(n)]
    top, bottom, left, right = 0, n - 1, 0, n - 1
    num = 1
    while top <= bottom and left <= right:
        for col in range(left, right + 1):
            matrix[top][col] = num
            num += 1
        top += 1
        for row in range(top, bottom + 1):
            matrix[row][right] = num
            num += 1
        right -= 1
        if top <= bottom:
            for col in range(right, left - 1, -1):
                matrix[bottom][col] = num
                num += 1
            bottom -= 1
        if left <= right:
            for row in range(bottom, top - 1, -1):
                matrix[row][left] = num
                num += 1
            left += 1
    for row in matrix:
        print(" ".join(str(x).rjust(3) for x in row))

spiral_pattern(4)
```

**Q23.** Use reduce() to find the second largest number in a list without sorting.

```python
from functools import reduce

def second_largest(nums):
    def helper(acc, x):
        largest, second = acc
        if x > largest:
            return (x, largest)
        elif largest > x > second:
            return (largest, x)
        return (largest, second)

    first, second = reduce(helper, nums[2:], (max(nums[0], nums[1]), min(nums[0], nums[1])))
    return second

print(second_largest([10, 20, 4, 45, 99, 99, 3]))
```

```python
# DANGEROUS version: default argument list is created ONCE and shared across calls
def add_item_bad(item, items=[]):
    items.append(item)
    return items

print(add_item_bad("a"))  # ['a']
print(add_item_bad("b"))  # ['a', 'b']  <-- unexpected! previous call's list persisted

# SAFE version: use None as default and create a new list each call
def add_item_good(item, items=None):
    if items is None:
        items = []
    items.append(item)
    return items

print(add_item_good("a"))  # ['a']
print(add_item_good("b"))  # ['b']  <-- as expected
```

**Q30.** Create a custom iterator class that returns Fibonacci numbers up to n terms.

```python
class FibonacciIterator:
    def __init__(self, n):
        self.n = n
        self.count = 0
        self.a, self.b = 0, 1

    def __iter__(self):
        return self

    def __next__(self):
        if self.count >= self.n:
            raise StopIteration
        result = self.a
        self.a, self.b = self.b, self.a + self.b
        self.count += 1
        return result

for num in FibonacciIterator(10):
    print(num, end=" ")
```

**Q36.** Write a recursive function to solve the Josephus problem for n people and step k.

```python
def josephus(n, k):
    # Base case: only one person left, they are the survivor (0-indexed)
    if n == 1:
        return 0
    # Recurrence: position in smaller circle of n-1 mapped back to n
    return (josephus(n - 1, k) + k) % n

n, k = 7, 3
# Convert to 1-indexed position for readability
print("Survivor position (1-indexed):", josephus(n, k) + 1)
```

**Q37.** Write a decorator that caches (memoizes) results of a recursive Fibonacci function.

```python
def memoize(func):
    cache = {}
    def wrapper(*args):
        if args not in cache:
            cache[args] = func(*args)
        return cache[args]
    return wrapper

@memoize
def fib(n):
    if n <= 1:
        return n
    return fib(n - 1) + fib(n - 2)

print([fib(i) for i in range(15)])
```

```python
class InfiniteCycle:
    def __init__(self, items):
        self.items = items
        self.index = 0

    def __iter__(self):
        return self

    def __next__(self):
        if not self.items:
            raise StopIteration
        item = self.items[self.index % len(self.items)]
        self.index += 1
        return item

cycler = InfiniteCycle(["a", "b", "c"])
result = [next(cycler) for _ in range(8)]  # take first 8 values manually
print(result)
```

**Q39.** Write a generator that yields prime numbers indefinitely (infinite generator).

```python
def prime_generator():
    def is_prime(num):
        if num < 2:
            return False
        for i in range(2, int(num ** 0.5) + 1):
            if num % i == 0:
                return False
        return True

    num = 2
    while True:
        if is_prime(num):
            yield num
        num += 1

gen = prime_generator()
first_15_primes = [next(gen) for _ in range(15)]
print(first_15_primes)
```

**Q40.** Write a decorator that restricts a function to being called no more than N times, raising an error after.

```python
def limit_calls(max_calls):
    def decorator(func):
        call_count = [0]
        def wrapper(*args, **kwargs):
            if call_count[0] >= max_calls:
                raise RuntimeError(f"{func.__name__} can only be called {max_calls} times")
            call_count[0] += 1
            return func(*args, **kwargs)
        return wrapper
    return decorator

@limit_calls(3)
def greet(name):
    return f"Hello, {name}!"

print(greet("Alice"))
print(greet("Bob"))
print(greet("Charlie"))
try:
    print(greet("David"))
except RuntimeError as e:
    print("Error:", e)
```

**Q41.** Given raw server log lines, use a generator pipeline (map+filter chained) to yield only ERROR lines from the last hour, without loading the whole file into memory.

```python
from datetime import datetime, timedelta

def log_line_generator(lines):
    # Simulates reading a file lazily, one line at a time
    for line in lines:
        yield line

def parse_line(line):
    parts = line.split(" ", 2)
    timestamp = datetime.strptime(parts[0] + " " + parts[1], "%Y-%m-%d %H:%M:%S")
    level_and_msg = parts[2]
    return timestamp, level_and_msg

def error_pipeline(lines, now, window=timedelta(hours=1)):
    parsed = map(parse_line, log_line_generator(lines))
    recent_errors = filter(
        lambda t: t[1].startswith("ERROR") and (now - t[0]) <= window,
        parsed
    )
    for timestamp, msg in recent_errors:
        yield f"{timestamp} {msg}"

raw_logs = [
    "2026-08-01 09:00:00 ERROR Disk full",
    "2026-08-01 09:20:00 INFO Started job",
    "2026-08-01 09:40:00 ERROR Connection timeout",
    "2026-08-01 07:00:00 ERROR Old error (too old)",
]
current_time = datetime.strptime("2026-08-01 09:45:00", "%Y-%m-%d %H:%M:%S")

for entry in error_pipeline(raw_logs, current_time):
    print(entry)
```

**Q44.** Implement a rate limiter using a decorator that allows a function to run only once every N seconds (using time module).

```python
import time

def rate_limit(seconds):
    def decorator(func):
        last_called = [0]
        def wrapper(*args, **kwargs):
            now = time.time()
            if now - last_called[0] < seconds:
                print(f"Rate limit hit: try again in {seconds - (now - last_called[0]):.2f}s")
                return None
            last_called[0] = now
            return func(*args, **kwargs)
        return wrapper
    return decorator

@rate_limit(2)
def say_hello():
    print("Hello!")

say_hello()  # runs
say_hello()  # blocked, called too soon
time.sleep(2)
say_hello()  # runs again after waiting
```

**Q46.** Implement a generator-based pipeline that reads a CSV of transactions and yields only fraudulent ones (amount > 3x the account's average).

```python
import csv
import io
from collections import defaultdict

csv_data = """account,amount
A1,100
A1,120
A1,90
A1,500
A2,50
A2,55
A2,300
"""

def read_transactions(csv_text):
    reader = csv.DictReader(io.StringIO(csv_text))
    for row in reader:
        yield row["account"], float(row["amount"])

def fraud_pipeline(csv_text):
    # First pass: compute average per account (needs full data for the average)
    transactions = list(read_transactions(csv_text))
    totals = defaultdict(list)
    for acc, amt in transactions:
        totals[acc].append(amt)
    averages = {acc: sum(amts) / len(amts) for acc, amts in totals.items()}

    # Second pass: generator that yields only fraudulent transactions
    for acc, amt in transactions:
        if amt > 3 * averages[acc]:
            yield (acc, amt, averages[acc])

for acc, amt, avg in fraud_pipeline(csv_data):
    print(f"Fraud flagged: account={acc}, amount={amt}, avg={avg:.2f}")
```

**Q47.** Write a custom iterator that implements a sliding window of size k over a list (yields overlapping sub-lists).

```python
class SlidingWindow:
    def __init__(self, data, k):
        self.data = data
        self.k = k
        self.index = 0

    def __iter__(self):
        return self

    def __next__(self):
        if self.index + self.k > len(self.data):
            raise StopIteration
        window = self.data[self.index:self.index + self.k]
        self.index += 1
        return window

for w in SlidingWindow([1, 2, 3, 4, 5, 6], 3):
    print(w)
```

**Q48.** Using polymorphism, design PaymentMethod base class with CreditCard, UPI, Wallet subclasses each having a pay() method with different validation logic; process a mixed list of payments and report failures.

```python
class PaymentMethod:
    def __init__(self, amount):
        self.amount = amount

    def pay(self):
        raise NotImplementedError("Subclasses must implement pay()")


class CreditCard(PaymentMethod):
    def __init__(self, amount, card_number):
        super().__init__(amount)
        self.card_number = card_number

    def pay(self):
        if len(self.card_number) != 16:
            raise ValueError("Invalid credit card number")
        return f"Paid {self.amount} via Credit Card ending {self.card_number[-4:]}"


class UPI(PaymentMethod):
    def __init__(self, amount, upi_id):
        super().__init__(amount)
        self.upi_id = upi_id

    def pay(self):
        if "@" not in self.upi_id:
            raise ValueError("Invalid UPI ID")
        return f"Paid {self.amount} via UPI ({self.upi_id})"


class Wallet(PaymentMethod):
    def __init__(self, amount, balance):
        super().__init__(amount)
        self.balance = balance

    def pay(self):
        if self.amount > self.balance:
            raise ValueError("Insufficient wallet balance")
        return f"Paid {self.amount} via Wallet, remaining balance {self.balance - self.amount}"


payments = [
    CreditCard(500, "1234567812345678"),
    CreditCard(200, "123"),               # invalid card number -> failure
    UPI(150, "user@bank"),
    UPI(100, "invalid-upi"),              # invalid UPI -> failure
    Wallet(50, 100),
    Wallet(300, 100),                     # insufficient balance -> failure
]

successes, failures = [], []
for payment in payments:
    try:
        successes.append(payment.pay())
    except ValueError as e:
        failures.append(f"{type(payment).__name__}: {e}")

print("Successful payments:")
for s in successes:
    print(" -", s)

print("\nFailed payments:")
for f in failures:
    print(" -", f)
```

**Q49.** Write a decorator that logs function arguments and return value to a file only when the function raises an exception.

```python
import functools

def log_on_exception(log_file="error_log.txt"):
    def decorator(func):
        @functools.wraps(func)
        def wrapper(*args,**kwargs):
            try:
                return func(*args, **kwargs)
            except Exception as e:
                with open(log_file, "a") as f:
                    f.write(
                        f"Function: {func.__name__}, args={args}, kwargs={kwargs}, "
                        f"exception={e}\n"
                    )
                raise
        return wrapper
    return decorator

@log_on_exception()
def divide(a, b):
    return a / b

try:
    divide(10, 0)
except ZeroDivisionError:
    print("Caught exception, check error_log.txt for details")

with open("error_log.txt") as f:
    print(f.read())
```

**Q50.** Given a string representing nested brackets and text (e.g. "a(b(c)d)e"), write a recursive function to extract content at a given nesting depth.

```python
def extract_at_depth(s, target_depth):
    result = []
    current_depth = 0
    buffer = []

    def helper(index, depth):
        nonlocal buffer
        while index < len(s):
            ch = s[index]
            if ch == "(":
                if depth + 1 == target_depth:
                    buffer = []
                index = helper(index + 1, depth + 1)
                if depth + 1 == target_depth:
                    result.append("".join(buffer))
            elif ch == ")":
                return index + 1
            else:
                if depth == target_depth:
                    buffer.append(ch)
                index += 1
        return index

    helper(0, 0)
    return result

text ="a(b(c)d)e"
print("Depth 1 content:", extract_at_depth(text, 1))
print("Depth 2 content:", extract_at_depth(text, 2))
```
