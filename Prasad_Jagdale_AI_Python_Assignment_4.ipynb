6. Create a class Wallet with a private balance; block direct modification but allow deposit()/withdraw() only.


```python
class Wallet:
    def __init__(self, balance=0):
        self.__balance = balance

    @property
    def balance(self):
        return self.__balance

    def deposit(self, amount):
        if amount <= 0:
            raise ValueError("Deposit must be positive.")
        self.__balance += amount

    def withdraw(self, amount):
        if amount <= 0 or amount > self.__balance:
            raise ValueError("Invalid withdrawal.")
        self.__balance -= amount

wallet = Wallet(100)
wallet.deposit(50)
wallet.withdraw(30)
print("Balance:", wallet.balance)

try:
    wallet.__balance = 1000
except Exception as e:
    print(e)

print("Actual balance remains:", wallet.balance)
```

12. Implement area() and perimeter() in Circle and Rectangle; try instantiating Shape directly and catch the error.


```python
from abc import ABC, abstractmethod
import math

class Shape(ABC):
    @abstractmethod
    def area(self):
        pass

    @abstractmethod
    def perimeter(self):
        pass

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return math.pi * self.radius ** 2

    def perimeter(self):
        return 2 * math.pi * self.radius

class Rectangle(Shape):
    def __init__(self, width, height):
        self.width = width
        self.height = height

    def area(self):
        return self.width * self.height

    def perimeter(self):
        return 2 * (self.width + self.height)

for obj in [Circle(3), Rectangle(4, 5)]:
    print("Area:", obj.area(), "Perimeter:", obj.perimeter())

try:
    Shape()
except TypeError as e:
    print("Cannot instantiate Shape:", e)
```

22. Write a class using @property where the setter recalculates a dependent attribute automatically.


```python
class Rectangle:
    def __init__(self, width, height):
        self.width = width
        self.height = height
        self._area = width * height

    @property
    def area(self):
        return self._area

    @property
    def width(self):
        return self._width

    @width.setter
    def width(self, value):
        if value < 0:
            raise ValueError("Width cannot be negative.")
        self._width = value
        if hasattr(self, "_height"):
            self._area = self._width * self._height

    @property
    def height(self):
        return self._height

    @height.setter
    def height(self, value):
        if value < 0:
            raise ValueError("Height cannot be negative.")
        self._height = value
        if hasattr(self, "_width"):
            self._area = self._width * self._height

r = Rectangle(4, 5)
print("Area:", r.area)
r.width = 10
print("Updated area:", r.area)
```

23. Create a class Temperature with properties for both Celsius and Fahrenheit that stay in sync (changing one updates the other).


```python
class Temperature:
    def __init__(self, celsius=0):
        self.celsius = celsius

    @property
    def celsius(self):
        return self._celsius

    @celsius.setter
    def celsius(self, value):
        self._celsius = value

    @property
    def fahrenheit(self):
        return (self._celsius * 9 / 5) + 32

    @fahrenheit.setter
    def fahrenheit(self, value):
        self._celsius = (value - 32) * 5 / 9

t = Temperature(25)
print("Celsius:", t.celsius)
print("Fahrenheit:", t.fahrenheit)

t.fahrenheit = 212
print("After changing Fahrenheit:")
print("Celsius:", t.celsius)
print("Fahrenheit:", t.fahrenheit)
```

 42. Create a BankAccount with encapsulated balance and a transaction history list; overdraft allowed up to a limit only for "Premium" accounts (use inheritance to differentiate).


```python
class BankAccount:
    def __init__(self, balance=0):
        self._balance = balance
        self.transaction_history = []

    def deposit(self, amount):
        self._balance += amount
        self.transaction_history.append(f"Deposit: {amount}")

    def withdraw(self, amount):
        if amount > self._balance:
            raise ValueError("Overdraft not allowed.")
        self._balance -= amount
        self.transaction_history.append(f"Withdrawal: {amount}")

    @property
    def balance(self):
        return self._balance

class PremiumAccount(BankAccount):
    overdraft_limit = 10000

    def withdraw(self, amount):
        if amount > self._balance + self.overdraft_limit:
            raise ValueError("Overdraft limit exceeded.")

        self._balance -= amount
        self.transaction_history.append(f"Withdrawal: {amount}")

regular = BankAccount(5000)
premium = PremiumAccount(5000)

try:
    regular.withdraw(6000)
except ValueError as e:
    print("Regular:", e)

premium.withdraw(6000)
print("Premium balance:", premium.balance)
print("History:", premium.transaction_history)
```

43. Create an abstract class Employee with calculate_salary(); implement Intern (stipend), FullTime (fixed+bonus), Contractor (hourly, capped at 160 hrs) — process a mixed payroll list and print total company payout.


```python
from abc import ABC, abstractmethod

class Employee(ABC):
    @abstractmethod
    def calculate_salary(self):
        pass

class Intern(Employee):
    def __init__(self, stipend):
        self.stipend = stipend

    def calculate_salary(self):
        return self.stipend

class FullTime(Employee):
    def __init__(self, fixed, bonus):
        self.fixed = fixed
        self.bonus = bonus

    def calculate_salary(self):
        return self.fixed + self.bonus

class Contractor(Employee):
    def __init__(self, hourly_rate, hours):
        self.hourly_rate = hourly_rate
        self.hours = min(hours, 160)

    def calculate_salary(self):
        return self.hourly_rate * self.hours

payroll = [
    Intern(15000),
    FullTime(60000, 10000),
    Contractor(500, 180)
]

total = sum(employee.calculate_salary() for employee in payroll)

for employee in payroll:
    print(type(employee).__name__, ":", employee.calculate_salary())

print("Total company payout:", total)
```

44. Create a class Playlist overriding __iter__ and __next__ to make it directly iterable (custom iterator protocol, not just __getitem__).


```python
class Playlist:
    def __init__(self, songs):
        self.songs = songs
        self.index = 0

    def __iter__(self):
        self.index = 0
        return self

    def __next__(self):
        if self.index >= len(self.songs):
            raise StopIteration

        song = self.songs[self.index]
        self.index += 1
        return song

playlist = Playlist(["Song A", "Song B", "Song C"])

for song in playlist:
    print(song)
```

 45. Create a class Student with a static method validating a roll number AND a classmethod that creates a Student object directly from a raw CSV line.


```python
class Student:
    def __init__(self, roll_no, name):
        self.roll_no = roll_no
        self.name = name

    @staticmethod
    def validate_roll(roll_no):
        return str(roll_no).isdigit() and int(roll_no) > 0

    @classmethod
    def from_csv(cls, line):
        roll_no, name = line.split(",")
        roll_no = roll_no.strip()
        name = name.strip()

        if not cls.validate_roll(roll_no):
            raise ValueError("Invalid roll number.")

        return cls(int(roll_no), name)

student = Student.from_csv("101, Prasad")
print(student.roll_no, student.name)
```

 46. Create a class SparseMatrix overriding __getitem__/__setitem__ that only stores non-zero values internally (memory-efficient real-world use of dunder methods).


```python
class SparseMatrix:
    def __init__(self):
        self._data = {}

    def __getitem__(self, position):
        return self._data.get(position, 0)

    def __setitem__(self, position, value):
        if value == 0:
            self._data.pop(position, None)
        else:
            self._data[position] = value

matrix = SparseMatrix()

matrix[0, 0] = 5
matrix[1, 2] = 10
matrix[2, 2] = 0

print("matrix[0, 0]:", matrix[0, 0])
print("matrix[1, 2]:", matrix[1, 2])
print("matrix[2, 2]:", matrix[2, 2])
print("Stored non-zero values:", matrix._data)
```

```python
class Booking:
    VALID_TRANSITIONS = {
        "Pending": {"Confirmed", "Cancelled"},
        "Confirmed": {"Cancelled"},
        "Cancelled": set()
    }

    def __init__(self):
        self.__status = "Pending"

    @property
    def status(self):
        return self.__status

    @status.setter
    def status(self, new_status):
        if new_status not in self.VALID_TRANSITIONS:
            raise ValueError("Invalid status.")

        if new_status not in self.VALID_TRANSITIONS[self.__status]:
            raise ValueError(f"Invalid transition: {self.__status} -> {new_status}")

        self.__status = new_status

booking = Booking()
print("Status:", booking.status)

booking.status = "Confirmed"
print("Status:", booking.status)

booking.status = "Cancelled"
print("Status:", booking.status)

try:
    booking.status = "Pending"
except ValueError as e:
    print(e)
```

 48. Design a Shape hierarchy with polymorphism where describe() calls area() and perimeter(); add a Triangle subclass and validate triangle inequality in its constructor, raising an error otherwise.


```python
from abc import ABC, abstractmethod
import math

class Shape(ABC):
    @abstractmethod
    def area(self):
        pass

    @abstractmethod
    def perimeter(self):
        pass

    def describe(self):
        print("Area:", self.area())
        print("Perimeter:", self.perimeter())

class Triangle(Shape):
    def __init__(self, a, b, c):
        if a + b <= c or a + c <= b or b + c <= a:
            raise ValueError("Invalid triangle: triangle inequality violated.")
        self.a, self.b, self.c = a, b, c

    def area(self):
        s = (self.a + self.b + self.c) / 2
        return math.sqrt(s * (s - self.a) * (s - self.b) * (s - self.c))

    def perimeter(self):
        return self.a + self.b + self.c

triangle = Triangle(3, 4, 5)
triangle.describe()

try:
    Triangle(1, 2, 5)
except ValueError as e:
    print(e)
```

 49. Create a class ConnectionPool using constructor to acquire a limited resource (max N connections class-wide) and destructor to release it; raise an error if pool is exhausted.


```python
class ConnectionPool:
    max_connections = 2
    active_connections = 0

    def __init__(self):
        if ConnectionPool.active_connections >= ConnectionPool.max_connections:
            raise RuntimeError("Connection pool exhausted.")

        ConnectionPool.active_connections += 1
        self.acquired = True
        print("Connection acquired.")

    def __del__(self):
        if getattr(self, "acquired", False):
            ConnectionPool.active_connections -= 1
            self.acquired = False
            print("Connection released.")

c1 = ConnectionPool()
c2 = ConnectionPool()

try:
    c3 = ConnectionPool()
except RuntimeError as e:
    print(e)

del c1
del c2

print("Active connections:", ConnectionPool.active_connections)
```

 50. Create a class Employee with a class-level static counter tracking active employees only (destructor decrements it on termination), and a classmethod to report the current active count.


```python
class Employee:
    active_count = 0

    def __init__(self, name):
        self.name = name
        Employee.active_count += 1

    def __del__(self):
        Employee.active_count -= 1

    @classmethod
    def get_active_count(cls):
        return cls.active_count

e1 = Employee("A")
e2 = Employee("B")
print("Active employees:", Employee.get_active_count())

del e1
print("After deleting e1:", Employee.get_active_count())

del e2
print("After deleting e2:", Employee.get_active_count())
```
