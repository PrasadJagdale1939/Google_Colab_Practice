```python
# Central limit Theorem:
# When multiple samples are drawn from a population and the  sample means are calculated ,
# the distribution of the sample  means would be normally distributed Irrespective of the
# distribution of the population.

# The mean of the Sample means would be Equal to population mean.the std deviation of the smaple
# means would be equal to the population std Deviation Divided by Square root of sample Size.
```
