### **Banking System Using Dictionary and objects**

```python
class Account:

    def __init__(self, account_number: int, name: str, balance: float):
        self.account_number = account_number
        self.name = name
        self.balance = balance

    def deposit(self, amount: float)-> bool:
        if amount <= 0:
            print("Invalid amount! Deposit amount must be positive.")
            return False
        self.balance += amount
        print(f"₹{amount} deposited successfully.")
        print(f"Current Balance: ₹{self.balance}")
        return True

    def withdraw(self, amount: float)-> bool:
        if amount <= 0:
            print("Invalid amount! Withdrawal amount must be positive.")
            return False
        if self.balance - amount < 3000:
            print("Withdrawal not allowed!")
            print("Minimum balance of ₹3000 must be maintained.")
            print(f"Current Balance: ₹{self.balance}")
            return False

        self.balance -=amount
        print("Withdrawal successful!")
        print(f"Amount Withdrawn: ₹{amount}")
        print(f"Remaining Balance: ₹{self.balance}")
        return True

    def display_details(self):
        print("\n--- ACCOUNT DETAILS ---")
        print(f"Account Number : {self.account_number}")
        print(f"Account Holder : {self.name}")
        print(f"Balance        : ₹{self.balance}")


def display_menu():
    print("\n===== BANKING SYSTEM =====")
    print("1. Open New Account")
    print("2. Deposit Money")
    print("3. Withdraw Money")
    print("4. Account-to-Account Transfer")
    print("5. Check Balance")
    print("6. Exit")


def main():
    bank = {}
    next_account_number= 20260001

    while True:
        display_menu()
        try:
            choice = int(input("Enter your choice: ").strip())
        except ValueError:
            print("Invalid input! Please enter a number between 1 and 6.")
            continue

    #1. Add Account
        if choice ==1:
            name= input("Enter Account Holder Name: ").strip()
            try:
                initial_deposit= float(input("Enter Initial Deposit: ").strip())
            except ValueError:
                print("Invalid amount entered. Account creation failed.")
                continue

            if initial_deposit < 5000:
                print("Account cannot be opened.")
                print("Minimum initial deposit is ₹5000.")
            else:
                new_acc = Account(next_account_number, name, initial_deposit)
                bank[next_account_number] = new_acc
                print("Account created successfully!")
                print(f"Account Number: {next_account_number}")
                print(f"Current Balance: ₹{initial_deposit}")
                next_account_number += 1

        # 2. Deposit Money
        elif choice== 2:
            try:
                acc_no =int(input("Enter Account Number: ").strip())
            except ValueError:
                print("Invalid account number.")
                continue

            if acc_no not in bank:
                print("Account does not exist.")
                continue

            try:
                amount = float(input("Enter Deposit Amount: ").strip())
                bank[acc_no].deposit(amount)
            except ValueError:
                print("Invalid amount.")

        # 3. Withdraw Money
        elif choice == 3:
            try:
                acc_no =int(input("Enter Account Number: ").strip())
            except ValueError:
                print("Invalid account number.")
                continue

            if acc_no not in bank:
                print("Account does not exist.")
                continue

            try:
                amount =float(input("Enter Withdrawal Amount: ").strip())
                bank[acc_no].withdraw(amount)
            except ValueError:
                print("Invalid amount.")

        # 4. Account-to-Account Transfer
        elif choice== 4:
            try:
                src_acc_no =int(input("Enter Source Account Number: ").strip())
            except ValueError:
                print("Invalid account number.")
                continue

            if src_acc_no not in bank:
                print("Source account does not exist.")
                print("Transfer cancelled.")
                continue

            try:
                dest_acc_no =int(input("Enter Destination Account Number: ").strip())
            except ValueError:
                print("Invalid account number.")
                continue

            if dest_acc_no not in bank:
                print("Destination account does not exist.")
                print("Transfer cancelled.")
                continue

            if src_acc_no == dest_acc_no:
                print("Source and Destination accounts cannot be the same.")
                continue

            try:
                transfer_amount =float(input("Enter Transfer Amount: ").strip())
            except ValueError:
                print("Invalid amount.")
                continue

            if transfer_amount <= 0:
                print("Transfer amount must be positive.")
                continue

            src_acc = bank[src_acc_no]
            dest_acc = bank[dest_acc_no]

            # Atomic validation
            if src_acc.balance - transfer_amount < 3000:
                print("Transfer not allowed!")
                print("Minimum balance of ₹3000 must be maintained.")
                print(f"Current Source Balance: ₹{src_acc.balance}")
            else:
                src_acc.balance -= transfer_amount
                dest_acc.balance += transfer_amount
                print("Transfer successful!")
                print(f"₹{transfer_amount} transferred successfully.")
                print(f"Source Account     : {src_acc.account_number} | Source Balance     : ₹{src_acc.balance}")
                print(f"Destination Account: {dest_acc.account_number} | Destination Balance: ₹{dest_acc.balance}")

        # 5. Check Balance
        elif choice == 5:
            try:
                acc_no = int(input("Enter Account Number: ").strip())
            except ValueError:
                print("Invalid account number.")
                continue

            if acc_no in bank:
                bank[acc_no].display_details()
            else:
                print("Account does not exist.")

        # 6. Exit
        elif choice == 6:
            print("Thank you for using our Banking System!")
            break

        else:
            print("Invalid choice! Please select an option from 1 to 6.")


if __name__ == "__main__":
  main()
```
