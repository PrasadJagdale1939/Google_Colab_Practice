```python
items = {
    101: ["Tea", 20.0],
    102: ["Coffee", 40.0],
    103: ["Maggi", 60.0],
    104: ["Bun Maska", 50.0],
    105: ["Water", 20.0]
}

bill = {}

print("========== CAFE BILLING SYSTEM ==========")
print(f"{'Sr':<7}{'items':<20}  {'Price(₹)':>6}")
print("="*41)
for i in items:
  item_name = items[i][0]
  price_name =items[i][1]
  sr_no=i

  print(f"{sr_no:<7}{item_name:<20}  ₹{price_name:>5}")

print()


while True:
    item_no= int(input("Enter item's Serial number or 0 to get Bill: "))

    if item_no==0:
        break

    elif item_no in items:
        name_item= items[item_no][0]
        qty= int(input(f"Enter quantity for {name_item}: "))

        if qty<=0:
            print("Invalid quantity.")
            print("Please enter a quantity greater than 0 .")


        else:
            name= items[item_no][0]
            cost= items[item_no][1]

            if item_no in bill:
                bill[item_no][2] = bill[item_no][2] + qty
                bill[item_no][3] = bill[item_no][2] * cost
                print(name + " quantity updated.")
            else:
                total = cost*qty
                bill[item_no]= [name, cost, qty, total]
                print(name +" added to bill.")
            print()
    else:
        print("Invalid item number.")
        print("Please select a valid item.")
        print( )

print("================ FINAL BILL ================")
print()
print(f"{'Item':<15}{'Qty':>6}{'Rate':>12}{'Total':>12}")
print("-"*48)

subtotal= 0.0

for i in bill:
    n= bill[i][0]
    c= bill[i][1]
    q= bill[i][2]
    t= bill[i][3]

    print(f"{n:<15}{q:>6}{c:>10.2f}₹{t:>10.2f}")
    subtotal += t


print("-"*48)
print()
print(f"Subtotal:{'':<27}₹{subtotal:.2f}")
print()

gst = subtotal * 2.5/100
cgst = subtotal*2.5 / 100
total_tax = gst+cgst
total_payable = subtotal + total_tax

print(f"GST @ 2.5%:{'':<27}₹{gst:.2f}")
print(f"CGST @ 2.5%:{'':<26}₹{cgst:.2f}")
print("-"*48)
print()
print(f"Total Tax:{'':<30}₹{total_tax:.2f}")
print()
print(f"TOTAL PAYABLE:{'':<25}₹{total_payable:.2f}")
print("-"*48)
print()
print(f"{'Thank You!':>30}")
print(f"{'Visit Again!':>31}")
```
