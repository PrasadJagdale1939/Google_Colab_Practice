```python
paragraph= input("Enter The paragraph:")

total_chars =len(paragraph)

chars_no_spaces= 0
for ch in paragraph:
    if ch != " ":
        chars_no_spaces += 1

raw_words=paragraph.split()

words= list()
for w in raw_words:
    cleaned = ""
    for ch in w:
        if ch.isalpha() or ch.isdigit():
            cleaned = cleaned + ch
    if cleaned != "":
        words.append(cleaned)

total_words= len(words)


total_sentences = 0
for ch in paragraph:
    if ch== "." or ch == "!" or ch == "?":
        total_sentences+= 1


vowels = 0
consonants = 0
digits = 0
specials = 0

vowel_letters ="aeiouAEIOU"

for ch in paragraph:
    if ch.isalpha():
        if ch in vowel_letters:
            vowels += 1
        else:
            consonants += 1
    elif ch.isdigit():
        digits += 1
    elif ch == " ":
        pass
    else:
        specials += 1


lower_words = []
for w in words:
    lower_words.append(w.lower())

most_common_word = ""
most_common_freq = 0

for w in lower_words:
    count= 0
    for w2 in lower_words:
        if w2== w:
            count+= 1
    if count > most_common_freq:
        most_common_freq= count
        most_common_word =w


longest_word = ""
for w in words:
    if len(w) > len(longest_word):
        longest_word = w

longest_word_len =len(longest_word)


print()
print(f"Total characters : {total_chars}")
print(f"Characters without spaces: {chars_no_spaces}")
print(f"Total words : {total_words}")
print(f"Total sentences : {total_sentences}")
print(f"Vowels : {vowels}")
print(f"Consonants : {consonants}")
print(f"Digits : {digits}")
print(f"Special characters : {specials}")
print(f"Most frequent word : {most_common_word}")
print(f"Frequency of most common word : {most_common_freq}")
print(f"Longest word : {longest_word}")
print(f"Length of longest word : {longest_word_len}")
```
