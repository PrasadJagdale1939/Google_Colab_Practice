```python
class Solution:
    def pushZerosToEnd(arr):
        # insert_pos keeps track of where the next non-zero element should go
        insert_pos = 0

        for i in range(len(arr)):
            if arr[i] != 0:
                # Swap the current non-zero element with the element at insert_pos
                arr[insert_pos], arr[i] = arr[i], arr[insert_pos]

                # Move our insertion pointer forward
                insert_pos += 1

        return arr
    a=[1,0,5,2,0,3,0,4,0]
    b=pushZerosToEnd(a)
    print(b)
```
