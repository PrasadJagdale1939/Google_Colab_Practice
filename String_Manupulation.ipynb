```python
items = {
    101: ["Tea", 20.0],
    102: ["Coffee", 40.0],
    103: ["Maggi", 60.0],
    104: ["Bun Maska", 50.0],
    105: ["Water", 20.0]
}

bill = {}

print("========== CAFE BILLING SYSTEM ==========")
print(f"{'Sr':<7}{'items':<20}  {'Price(₹)':>6}")
print("="*41)
for i in items:
  item_name = items[i][0]
  price_name =items[i][1]
  sr_no=i

  print(f"{sr_no:<7}{item_name:<20}  ₹{price_name:>5}")
print()


while True:
    item_no= int(input("Enter item's Serial number: "))

    if item_no == 0:
        break

    elif item_no in items:
        qty= int(input("Enter quantity for item: "))

        if qty <=0:
            print("Invalid quantity.")
            print("Please enter a quantity greater than 0 .")


        else:
            name= items[item_no][0]
            cost= items[item_no][1]

            if item_no in bill:
                bill[item_no][2] = bill[item_no][2] + qty
                bill[item_no][3] = bill[item_no][2] * cost
                print(name + " quantity updated.")
            else:
                total = cost * qty
                bill[item_no] = [name, cost, qty, total]
                print(name + " added to bill.")
            print()
    else:
        print("Invalid item number.")
        print("Please select a valid item.")
        print()

print("================ FINAL BILL ================")
print()
print(f"{'Item':<15}{'Qty':>6}{'Rate':>12}{'Total':>12}")
print("------------------------------------------------")

subtotal = 0.0

for item_no in bill:
    name = bill[item_no][0]
    cost = bill[item_no][1]
    qty = bill[item_no][2]
    total = bill[item_no][3]

    print(f"{name:<15}{qty:>6}{'₹' + format(cost, '.2f'):>12}{'₹' + format(total, '.2f'):>12}")
    subtotal = subtotal + total

print("------------------------------------------------")
print()
print(f"{'Subtotal:':<30}{'₹' + format(subtotal, '.2f'):>18}")
print()

gst= subtotal * 2.5 / 100
cgst= subtotal * 2.5 / 100
total_tax = gst + cgst
total_payable = subtotal + total_tax

print(f"{'GST @ 2.5%:':<30}{'₹' + format(gst, '.2f'):>18}")
print(f"{'CGST @ 2.5%:':<30}{'₹' + format(cgst, '.2f'):>18}")
print("------------------------------------------------")
print()
print(f"{'Total Tax:':<30}{'₹' + format(total_tax, '.2f'):>18}")
print()
print(f"{'TOTAL PAYABLE:':<30}{'₹' + format(total_payable, '.2f'):>18}")
print("------------------------------------------------")
print()
print(f"{'Thank You!':>30}")
print(f"{'Visit Again!':>31}")
```

```python
d={1:"one",2:"two",3:"three",4:"four"}

d[1]="one"  #Add or change one value
print(d)

print(d[3])  #Get one value using a key

print(d.get(4))   #Get one value using a key

print(d.keys())   #Get all keys

print(d.values())   #Get all values

print(d.items())    #Get key-value pairs

d[5]="Five"   #Add or change one value
print(d)

d.update({6:"six",7:"Seven"})  #Add or modify multiple keys
print(d)

d.setdefault(0,"ZERO")
print(d)

returned =d.pop(7) #Delete specified key and return value
print(d)
print(returned)

poped=d.popitem()  #Delete last inserted pair and return Key,Values
print(d)
print(poped)

exist= 4 in d   #Checks whether a key exists in the Dictionary.
print(exist)

not_exist = 4 not in d   #Checks whether a key does not exist.
print(not_exist)


d.clear()  #Delete all pairs
print(d)
```

```python
data = '''Python is easy to learn and Python is powerful.
Learning Python is fun and learning Python is useful.
Python helps students learn programming and Python helps students think.
Students learn Python because Python makes programming easy.
Python is easy, Python is useful, and Python is powerful.'''

# 1. Clean the text: remove punctuation and convert to lowercase
clean_data = data.replace('.', '').replace(',', '').lower()

# 2. Split the text into a list of words
words = sorted(clean_data.split())

# 3. Count the frequencies using a dictionary
word_freq = {}
for word in words:
    if word in word_freq:
        word_freq[word] += 1
    else:
        word_freq[word] = 1

# 4. Print the words and their frequencies
for word, count in word_freq.items():
    print(f"{word}: {count}")

```
